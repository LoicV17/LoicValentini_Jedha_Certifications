# 📇 Context

Fraud is a huge issue among financial institutions.  
In the EU in 2019, the European Central Bank estimated that fraudulent credit card transactions amounted to more than **€1 billion**! 😮

AI can really help solve this issue by detecting fraudulent payments in a very precise manner.  
This use case is now one of the most famous among Data Scientists.

However, even though we can build powerful algorithms, the hard part is putting them into **production**:  
👉 That means predicting fraudulent payments **in real-time** and responding appropriately.

---

# 🎯 Project Goals

The Business team has approached you with the following needs:

- ✅ Get **notified immediately** once a fraud is detected  
- ✅ Get a **daily report**, every morning, showing all payments and frauds from the **previous day**

---

## 📂 Available Data Sources

To complete this project, you will rely on two data sources:

### 1. **Fraudulent Payments Full Dataset**
A large historical dataset of payments labeled as **fraudulent or not**  
→ Use it to **train your algorithm**

### 2. **Real-time Payment API**
An API that retrieves **live payments**  
→ Use it for **real-time predictions**  
ℹ️ It gets updated **every minute**

# 🦮 Tips & Guidance

## 🔧 How to build your algorithm?

You can use any tool: `scikit-learn`, `TensorFlow`, API-based tools like **AmazonML**, or even **no-code** tools.  
If you don’t want to build the algorithm from scratch, use an external service.

> 🧠 Remember: build something **reusable**!  
> Using a model is not just calling `.predict()` 😉

---

## ⚠️ VERY IMPORTANT

> 🚨 The most important part is the **Data Pipeline**, **NOT** the ML Algorithm!

- The model is secondary  
- Your priority = a working, automated **pipeline**

---

## 🔰 Where to start?

Your project should **at least** include:

- ✅ Training an algorithm  
- ✅ Staging the model into **production**  
- ✅ Storing real-time data in a **database**

---

## 🔧 Minimum Architecture

The only minimal elements you must implement are:

- A component that **collects & stores** data  
- A component that **consumes & uses** the data  
- An **ETL** (or **ELT**) pipeline between them
