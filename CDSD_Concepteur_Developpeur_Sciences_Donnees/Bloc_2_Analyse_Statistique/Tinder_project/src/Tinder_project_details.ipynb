Description de l'entreprise 📇

Tinder est une application de rencontres en ligne et de réseau géosocial. Sur Tinder, les utilisateurs "swipent à droite" pour aimer ou "swipent à gauche" pour ne pas aimer les profils d'autres utilisateurs, qui incluent leurs photos, une courte biographie et une liste de leurs centres d'intérêt.

Tinder a été lancé par Sean Rad lors d'un hackathon organisé à l'incubateur Hatch Labs à West Hollywood en 2012.

En 2021, Tinder a enregistré plus de 65 milliards de matchs dans le monde.

Projet 🚧

L'équipe marketing a besoin d'aide pour un nouveau projet. Ils constatent une diminution du nombre de matchs et cherchent à comprendre ce qui rend les gens intéressés les uns par les autres.

Ils ont décidé de mener une expérience de speed dating avec des personnes ayant fourni beaucoup d'informations à Tinder sur elles-mêmes, pouvant éventuellement se refléter sur leur profil de rencontre sur l'application.

Tinder a ensuite rassemblé les données de cette expérience. Chaque ligne dans le jeu de données représente un speed date entre deux personnes et indique si chacune d'elles a secrètement accepté de partir pour un deuxième rendez-vous avec l'autre personne.

Portée de ce projet 🖼️

Les données ont été collectées auprès des participants à des événements expérimentaux de speed dating de 2002 à 2004. Lors des événements, les participants avaient un "premier rendez-vous" de quatre minutes avec chaque autre participant du sexe opposé. À la fin de leurs quatre minutes, les participants se demandaient s'ils aimeraient revoir leur rendez-vous. On leur demandait également de noter leur rendez-vous sur six attributs : Attractivité, Sincérité, Intelligence, Fun, Ambition et Intérêts partagés.

Le jeu de données comprend également des données de questionnaire collectées auprès des participants à différents moments du processus. Ces champs incluent : démographie, habitudes de rencontres, auto-perception sur des attributs clés, croyances sur ce que les autres trouvent précieux chez un partenaire et informations sur le mode de vie. Aides 🦮

Pour vous aider à réaliser ce projet, voici quelques conseils qui devraient vous aider :

Idées d'exploration des données :

Quels sont les attributs les moins désirables chez un partenaire masculin ? Est-ce différent pour les partenaires féminines ?
À quel point les gens pensent-ils que l'attractivité est importante dans la sélection d'un partenaire potentiel par rapport à son impact réel ?
Les intérêts partagés sont-ils plus importants qu'un background racial partagé ?
Les gens peuvent-ils prédire avec précision leur propre valeur perçue sur le marché des rencontres ?
En termes d'obtenir un deuxième rendez-vous, est-il préférable d'être le premier speed date de la soirée de quelqu'un ou le dernier ?
