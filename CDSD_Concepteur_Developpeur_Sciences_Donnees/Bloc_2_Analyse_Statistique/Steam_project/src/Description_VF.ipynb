Projet : La plateforme de jeux vidéo Steam 👾
⏱️ Durée estimée : 960 minutes
🎮 Entreprise : Steam

Description de l'entreprise 📇
Steam est une plateforme de distribution numérique de jeux vidéo développée par Valve. Elle a été lancée en septembre 2003 en tant que client logiciel pour fournir automatiquement des mises à jour pour les jeux de Valve. Elle s’est ensuite élargie à la distribution de jeux tiers fin 2005.
Steam offre diverses fonctionnalités, telles que la gestion des droits numériques (DRM), un système de matchmaking pour serveurs de jeux avec des mesures anti-triche (Valve Anti-Cheat), des fonctions de réseau social, et des services de streaming de jeux.
Le client Steam permet aussi l’automatisation des mises à jour, le stockage en cloud des sauvegardes, et l'accès à des fonctionnalités communautaires (messagerie directe, overlay en jeu, marché d’objets virtuels…).

Projet 🚧
Vous travaillez pour Ubisoft, un éditeur français de jeux vidéo.
Ubisoft souhaite sortir un nouveau jeu révolutionnaire !
Ils vous ont demandé de réaliser une analyse globale des jeux disponibles sur le marketplace de Steam afin de mieux comprendre l’écosystème vidéoludique et les tendances actuelles.

Objectifs 🎯
L’objectif final de ce projet est de comprendre quels facteurs influencent la popularité ou les ventes d’un jeu vidéo.
Mais votre manager souhaite aussi que vous profitiez de l’opportunité pour analyser plus globalement le marché du jeu vidéo.

Pour cela, vous devrez adopter différents niveaux d’analyse. Voici une liste d'exemples de questions intéressantes que vous pouvez explorer :

Analyse "macro"
Quel éditeur a publié le plus de jeux sur Steam ?

Quels sont les jeux les mieux notés ?

Y a-t-il des années avec plus de sorties ? Moins de sorties pendant la période Covid, par exemple ?

Comment les prix sont-ils répartis ? Y a-t-il beaucoup de jeux en réduction ?

Quelles sont les langues les plus représentées ?

Y a-t-il beaucoup de jeux interdits aux moins de 16/18 ans ?

Analyse des genres
Quels sont les genres les plus représentés ?

Certains genres ont-ils un meilleur ratio d’avis positifs/négatifs ?

Certains éditeurs ont-ils des genres préférés ?

Quels sont les genres les plus lucratifs ?

Analyse des plateformes
La majorité des jeux sont-ils disponibles sur Windows / Mac / Linux ?

Certains genres sont-ils préférentiellement disponibles sur certaines plateformes ?

➡️ Vous êtes libre de suivre ces axes d’analyse ou d’en choisir d’autres, tant que votre travail met en lumière des informations pertinentes et utiles. 🤓

Périmètre du projet 🖼️
Vous devrez utiliser Databricks et PySpark pour réaliser cette analyse exploratoire (EDA).
En particulier, vous utiliserez l’outil de visualisation intégré de Databricks pour créer vos graphiques.

📂 Le dataset est disponible dans le bucket S3 à l’adresse suivante :
s3://full-stack-bigdata-datasets/Big_Data/Project_Steam/steam_game_output.json

Aides 🦮
Quelques conseils pour vous guider dans le projet :

Pour adopter différents niveaux d’analyse, il peut être utile de créer plusieurs DataFrames.

Le jeu de données est semi-structuré avec un schéma imbriqué : les méthodes PySpark comme getField() et explode() pourront vous aider.

Il contient des champs texte et des dates : PySpark offre des fonctions utiles pour les manipuler efficacement 💡

N’hésitez pas à utiliser des fonctions d’agrégation et groupBy pour vos analyses segmentées.

Livrables 📬
Pour compléter ce projet, vous devrez fournir :

Un ou plusieurs notebooks incluant la manipulation de données avec PySpark et des visualisations à l’aide de l’outil de dashboard de Databricks.

Si Databricks indique que le notebook dépasse la taille maximale autorisée, divisez-le en plusieurs notebooks.

Collez les liens vers vos notebooks publiés dans votre dépôt GitHub, afin que le jury puisse les consulter facilement. 😌
